Tensorflow Implemnetation:

````python
import tensorflow as tf
import numpy as np
````

````python
# 1. Create constants
a = tf.constant(10)
b = tf.constant(20)
print("Addition:", (a + b).numpy())
````

````python
# 2. Create a 2D tensor
x = tf.constant([[1, 2], [3, 4]])
print("2D Tensor:\n",x.numpy())  #.numpy() is used when u only want matrix instead of shape and dtype 
````

````python
# 3. Arithmetic Operations
print("\nAddition:\n", (x + 2))
print("Subtraction:\n", (x - 2))
print("Multiplication:\n", (x * 2))
print("Division:\n", (x / 2))
print("Modulo:\n", (x % 2))
````

````python
# 4. Create zeros, ones and random tensors
print("\nZeros:\n", tf.zeros((2, 4)))
print("Ones:\n", tf.ones((3, 4)))
print("Random tensor:\n", tf.random.uniform((2, 2)))
````

````python
# 5. Check tensor properties
print("\nShape:", x.shape)
print("Rank:", tf.rank(x).numpy())
print("Data type:", x.dtype)
````

````python
# 6. Reshape tensor
y = tf.reshape(x, (4,))
print("\nReshaped tensor:", y.numpy())
````

````python

# 7. 2D matrix multiplication
m = tf.constant([[1, 2], [3, 4]])
n = tf.constant([[5, 6], [7, 8]])
print("\n2D Matrix multiplication:\n", tf.matmul(m, n))
````

````python

# 8. 3D tensor multiplication
p = tf.constant([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
q = tf.constant([
    [[1, 0], [0, 1]],
    [[2, 0], [0, 2]]
])
print("\n3D Tensor shape:", p.shape)
print("3D Matrix multiplication:\n", tf.matmul(p, q).numpy())
````

````python
# 9. Indexing and slicing,sum
print("\nFirst row:", x[0].numpy())
print("Element:", x[0, 1].numpy())
print("First column:", x[:, 0].numpy())
print("\nSum:", tf.reduce_sum(x).numpy())

````

````python
# 11. Transpose
print("\nTranspose:\n", tf.transpose(x).numpy())
````

````python
# optional
# 12. Convert TensorFlow tensor to NumPy
arr = x.numpy()
print("\nNumPy array:\n", arr)
print("Array type:", type(arr))

# 13. Convert NumPy array to TensorFlow tensor
arr2 = np.array([10, 20, 30])
t = tf.convert_to_tensor(arr2)
print("\nNumPy to TensorFlow:", t.numpy())

# 14. Compare tensors
print("\nIs each element greater than 2?\n", (x > 2).numpy())
````

Pytorch Implementation

````python
import torch

# 1. Create a tensor
x = torch.tensor([[1, 2], [3, 4]])
print("Tensor:\n", x)
````

````python
# 2. Create ones, zeros and random tensors
print("Ones:\n", torch.ones(2, 2))
print("Zeros:\n", torch.zeros(2, 2))
print("Random:\n", torch.rand(2, 2))
````

````python
# 3. Arithmetic operations
print("Addition:\n", x + 2)
print("Subtraction:\n", x - 2)
print("Multiplication:\n", x * 2)
print("Division:\n", x / 2)
print("Modulo:\n", x % 2)
````

````python

# 4. Tensor properties
print("Shape:", x.shape)
print("Dimensions:", x.ndim)
print("Data Type:", x.dtype)

# 5. Reshape, indexing and slicing
print("Reshape:\n", x.reshape(4))
print("First Row:", x[0])
print("First Column:", x[:, 0])

# 6. Sum, mean and maximum
print("Sum:", x.sum())
print("Mean:", x.float().mean())
print("Maximum:", x.max())
````

````python
# 7. Transpose and concatenation
print("Transpose:\n", x.T)
````

````python

# 8. 2D Matrix multiplication
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])
print("2D Multiplication:\n", torch.matmul(a, b))
````

````python

# 9. Check GPU availability
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)
print("GPU Available:", torch.cuda.is_available())
print("Device:", x.to(device).device)
````

````python

# 10. NumPy to PyTorch
arr = np.array([10, 20, 30])
tensor = torch.from_numpy(arr)
print("NumPy to PyTorch:\n", tensor)

# 11. PyTorch to NumPy
print("PyTorch to NumPy:\n", tensor.numpy())
````



### Q1. What is a tensor? Explain its types.

A tensor is a data structure used to store numbers in different dimensions.

- 0D: Scalar, e.g., `5`
- 1D: Vector, e.g., `[1, 2, 3]`
- 2D: Matrix, e.g., `[[1, 2], [3, 4]]`
- 3D: A collection of 2D matrices.

### Q2. What is deep learning?

Deep learning is a branch of machine learning that uses neural networks with multiple layers to learn patterns from data.

### Q3. What are TensorFlow and PyTorch? How are they different?

TensorFlow is an open-source machine learning framework developed by Google. PyTorch is an open-source machine learning framework that originated at Meta AI.

| TensorFlow                               | PyTorch                                     |
| ---------------------------------------- | ------------------------------------------- |
| Commonly uses the Keras API              | Uses a Python-friendly tensor API           |
| Tensor creation: `tf.constant()`         | Tensor creation: `torch.tensor()`           |
| Rank: `tf.rank(x)`                       | Rank: `x.ndim`                              |
| Matrix multiplication: `tf.matmul(a, b)` | Matrix multiplication: `torch.matmul(a, b)` |

Both support tensor operations, automatic differentiation, and GPU acceleration.

### Q4. What is the difference between an array and a tensor?

An array stores elements, while a tensor is a multidimensional data structure designed for numerical computation. Tensors in TensorFlow and PyTorch can also support GPU operations.

### Q5. Why do we use NumPy?

NumPy is a Python library for numerical calculations and array operations. It can exchange data with TensorFlow and PyTorch.

### Q6. Does our practical train a model?

No. Our practical demonstrates tensor operations, arithmetic, matrix multiplication, reshaping, indexing, and other library features. It does not train a model.

# Part 2: Tensor creation and properties

### Q7. How do you create tensors in TensorFlow and PyTorch?

TensorFlow:
x = tf.constant([[1, 2], [3, 4]])

PyTorch:
x = torch.tensor([[1, 2], [3, 4]])


Both create a 2D tensor.

### Q8. How do you create tensors containing zeros, ones, and random values?

TensorFlow:
tf.zeros((2, 2))tf.ones((2, 2))tf.random.uniform((2, 2))

PyTorch:
torch.zeros((2, 2))torch.ones((2, 2))torch.rand((2, 2))


These create tensors filled with zeros, ones, or random values.

### Q9. What are rank, shape, and data type?

- Rank: Number of dimensions.
- Shape: Size along each dimension.
- Data type: Type of values stored, such as integer or floating-point numbers.

For a 2D tensor of shape `(2, 2)`, the rank is `2`.

TensorFlow:
tf.rank(x)x.shapex.dtype

PyTorch:
x.ndimx.shapex.dtype


### Q10. What is reshaping?

Reshaping changes the shape of a tensor without changing its elements.

# TensorFlowtf.reshape(x, (4,))# PyTorchx.reshape(4)

A 2 × 2 tensor becomes a 1D tensor containing four elements.

### Q11. What are indexing and slicing?

Indexing accesses a specific element or row. Slicing selects a portion of a tensor.

x[0]      # First rowx[:, 0]   # First columnx[0, 1]   # First row, second column

These expressions work similarly in both frameworks.

### Q12. What is the difference between `ones()` and `ones_like()` in PyTorch?

- `torch.ones((2, 2))` creates a tensor with the specified shape.
- `torch.ones_like(x)` creates a tensor with the same shape and usually the same data type as `x`, filled with ones.

The equivalent `zeros()` and `zeros_like()` functions work similarly.

# Part 3: Arithmetic and matrix operations

### Q13. Which arithmetic operations can be performed on tensors?

Addition, subtraction, multiplication, division, and modulo.

| Operation      | Example |
| -------------- | ------- |
| Addition       | `x + 2` |
| Subtraction    | `x - 2` |
| Multiplication | `x * 2` |
| Division       | `x / 2` |
| Modulo         | `x % 2` |

These perform element-wise operations when applied to a tensor and a scalar.

### Q14. What is the difference between element-wise multiplication and matrix multiplication?

Element-wise multiplication multiplies corresponding elements. Matrix multiplication multiplies rows by columns according to mathematical rules.


x * 2           # Element-wise scalar multiplicationtf.matmul(a, b) # TensorFlow matrix multiplicationtorch.matmul(a, b) # PyTorch matrix multiplication

### Q15. What is the difference between 2D and 3D matrix multiplication?

- 2D matrix multiplication: Multiplies two matrices.
- 3D batch matrix multiplication: Performs matrix multiplication on batches of matrices.

A tensor of shape `(2, 2, 2)` contains two 2 × 2 matrices. Both TensorFlow's `tf.matmul()` and PyTorch's `torch.matmul()` support batched matrix multiplication.

### Q16. What are sum, mean, and maximum operations?

- Sum: Adds all elements.
- Mean: Calculates the average.
- Maximum: Returns the largest element.

For `[[1, 2], [3, 4]]`, the sum is `10`, the mean is `2.5`, and the maximum is `4`.

TensorFlow:
tf.reduce_sum(x)tf.reduce_mean(tf.cast(x, tf.float32))tf.reduce_max(x)


PyTorch:
torch.sum(x)torch.mean(x.float())torch.max(x)


### Q17. What is transpose?

Transpose exchanges rows and columns in a 2D tensor.

For example:

```
Original:       Transpose:
[[1, 2],        [[1, 3],
 [3, 4]]         [2, 4]]
```

In TensorFlow, use `tf.transpose(x)`. In PyTorch, use `x.T` for a 2D tensor.

### Q18. What is concatenation?

Concatenation joins tensors along an existing dimension.

TensorFlow:
tf.concat([x, x], axis=0)

PyTorch:
torch.cat((x, x), dim=0)


If `x` has shape `(2, 2)`, the result has shape `(4, 2)`.

# Part 4: NumPy, GPU, and advanced concepts

### Q19. How do you convert between NumPy arrays and tensors?

NumPy to TensorFlow:
tf.convert_to_tensor(arr)

NumPy to PyTorch:
torch.from_numpy(arr)

TensorFlow to NumPy:
x.numpy()

PyTorch to NumPy:
tensor.numpy()

The PyTorch conversion works directly for a compatible CPU tensor that does not require gradients. A GPU tensor must first be moved to the CPU.

### Q20. What is GPU acceleration? What are CUDA and `device`?

GPU acceleration uses a graphics processing unit to perform computations efficiently.
CUDA is NVIDIA's parallel computing platform. PyTorch can use compatible NVIDIA GPUs through CUDA.

torch.cuda.is_available()

This checks whether CUDA GPU computation is available to PyTorch
device = torch.device(    "cuda" if torch.cuda.is_available() else "cpu")x.to(device)

This selects an available CUDA device or falls back to the CPU, then moves the tensor to that device.

### Q21. What is the difference between normal and in-place operations in PyTorch?

- `t + 3` creates a new result and leaves `t` unchanged.
- `t.add_(3)` modifies `t` directly.

The underscore in `add_()` indicates an in-place operation.

### Q22. What are `tf.constant()` and `tf.Variable()`?

- `tf.constant()` creates a constant tensor whose value cannot be reassigned through TensorFlow operations.
- `tf.Variable()` creates a variable whose value can be updated, commonly for model parameters.

### Q23. What is automatic differentiation?

Automatic differentiation calculates derivatives of mathematical operations. TensorFlow provides `tf.GradientTape`, while PyTorch uses its autograd system. These are useful for neural network training, but our practical does not use them.
