```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Lambda, Dense
```

```python
# Sentence
sentence = "the cat eats fish the dog eats meat"
words = sentence.split()
```

```python
# Create vocabulary
vocab = list(set(words))

w = {}
for i in range(len(vocab)):
    w[vocab[i]] = i

rev = {}
for i in w:
    rev[w[i]] = i
```

```python
# Create training data
X = []
y = []

for i in range(1, len(words)-1):
    X.append([w[words[i-1]], w[words[i+1]]])
    y.append(w[words[i]])

X = np.array(X)
y = np.array(y)
```

```python
# Build CBOW model
model = Sequential([
    Embedding(len(vocab), 10),
    Lambda(lambda x: tf.reduce_mean(x, axis=1)),
    Dense(len(vocab), activation="softmax")
])

model.compile(optimizer="adam",loss="sparse_categorical_crossentropy")
```

```python
# Train model
model.fit(X, y, epochs=500, verbose=0)
```

```python
# Predict missing word
context = ["cat", "fish"]

test = np.array([[w["cat"], w["fish"]]])

pred = model.predict(test, verbose=0)
p = np.argmax(pred)

print("Sentence: the cat ____ fish")
print("Predicted word:", rev[p])
```



### Q1. What is CBOW? What is Word2Vec, and what is the aim of this practical?

CBOW stands for Continuous Bag of Words. It is a Word2Vec model used in Natural Language Processing (NLP) to predict a target word from its surrounding context words. Word2Vec converts words into numerical vectors called embeddings. The aim of this practical is to implement a CBOW model using TensorFlow and Keras to predict a missing word from a sentence.

### Q2. What are the types of Word2Vec models? Differentiate between CBOW and Skip-gram.

There are two main types of Word2Vec models: CBOW and Skip-gram. CBOW predicts the target word using surrounding context words, whereas Skip-gram predicts surrounding context words using a target word. CBOW is generally faster to train, while Skip-gram can perform well for learning representations of less frequent words.

### Q3. What is a context window? How does your code select context words?

A context window determines how many words around a target word are used for prediction. In our code, the word before and the word after the current word are selected as context words. For example, in the sentence "the cat eats fish", the context words `cat` and `fish` can be used to predict `eats`.

### Q4. Which libraries are used in your program, and why?

Our program uses NumPy, TensorFlow, and Keras. NumPy is used to store training data in arrays. TensorFlow performs numerical computations and averages embedding vectors. Keras provides the `Sequential`, `Embedding`, `Lambda`, and `Dense` layers to build the neural network.

### Q5. What is vocabulary? What are `word_to_id` and `id_to_word`?

Vocabulary is the collection of unique words in the sentence. In our code, `sorted(set(words))` removes duplicate words and sorts the remaining words. The dictionary `word_to_id` assigns a unique numerical ID to each word, while `id_to_word` converts a predicted ID back into its corresponding word.

### Q6. What are `X` and `y`? Why do we convert them into NumPy arrays?

In our program, `X` contains the IDs of the context words, and `y` contains the ID of the target word. The loop creates context-target pairs from the sentence. We convert `X` and `y` into NumPy arrays because they provide a convenient numerical format for training the neural network.

### Q7. What is an Embedding layer? Explain `Embedding(len(vocab), 10)`.

An Embedding layer converts each word ID into a dense numerical vector that the model learns during training. In `Embedding(len(vocab), 10)`, `len(vocab)` represents the number of unique words, and `10` represents the dimension of each word vector. These vectors help the model learn relationships between words.

### Q8. What is the purpose of the Lambda and Dense layers? Why is softmax used?

The Lambda layer calculates the average of the context-word embedding vectors to create a single context representation. The Dense layer uses this representation to calculate scores for all words in the vocabulary. The softmax activation function converts these scores into probabilities, allowing the model to select the most likely target word.

### Q9. What is `model.compile()`? Why are Adam and sparse categorical cross-entropy used?

The `model.compile()` function configures the model before training. Adam is the optimizer that updates the model's weights to reduce prediction errors. Sparse categorical cross-entropy is the loss function used to measure the difference between the predicted word distribution and the actual target word ID.

### Q10. What are epochs? What does `verbose=0` mean?

An epoch is one complete pass through the training dataset. In our program, `epochs=500` means the model processes the entire training dataset 500 times. The parameter `verbose=0` hides training progress and prevents the model from printing progress information during each epoch.

### Q11. How does your model predict a missing word? What do `model.predict()` and `np.argmax()` do?

The model receives the context words `cat` and `fish`, converts them into numerical IDs, and predicts a probability for every word in the vocabulary. The function `model.predict()` generates these predictions, and `np.argmax()` returns the index with the highest predicted score. Finally, `id_to_word` converts that index into the corresponding word. The intended prediction is `eats`, forming the sentence "the cat eats fish", although the small dataset means the result is not guaranteed.

### Q12. Is CBOW supervised or unsupervised learning? What is backpropagation?

CBOW is commonly considered a self-supervised learning technique because it generates its own training targets from the surrounding words in a text corpus. Backpropagation calculates how the model's parameters contribute to the prediction error. The optimizer then uses these gradients to update the weights and improve predictions.

### Q13. What are the applications, advantages, and limitations of CBOW?

CBOW is used in NLP applications such as word similarity, text classification, information retrieval, and learning word representations. Its main advantage is that it is generally computationally efficient. However, our implementation uses a very small corpus, so it may not learn reliable word relationships. It can predict only words included in its vocabulary.
